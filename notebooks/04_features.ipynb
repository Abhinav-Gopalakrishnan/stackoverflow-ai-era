### M4-06 Correlation and Leakage Check
We ran a correlation analysis against the target `target_resolved_7d` and checked for multicollinearity using Variance Inflation Factor (VIF).

#### Findings:
1. **Leakage**: No leaky variables (like score, view_count, accepted_date) were included in the features pipeline.
2. **Target Correlation**: The secondary target `target_answered_7d` correlates strongly (0.48) with `target_resolved_7d` as expected. Among actual features, `n_code_blocks` (0.07) and `has_error_text` (-0.05) are the strongest linear predictors.
3. **Multicollinearity (VIF)**: We identified perfect multicollinearity (VIF > 1e15) among the `question_word_*` columns. This is the classic **Dummy Variable Trap**. We must configure our OneHotEncoder to drop the first category (`drop='first'`) in our final model pipeline!
